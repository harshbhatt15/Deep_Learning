### NEXT WORD PREDICTOR THROUGH LSTM(LONG SHORT TERM MEMORY)

Importing the dependencies


```python
import os 
import json

import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense , Embedding , LSTM
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

spliting Data into Train_test_split

```python
train_data , test_data = train_test_split(df , test_size=0.2,random_state=2)
```

not spliting data into x_train and Y_train first i'll do preprocessing and then i will split into x_train and Y_train.


```python
print(train_data.shape)
print(test_data.shape)
```

Data Preprocessing

```python
# tokenize text data 
tokenizer = Tokenizer(num_words=5000) # convert words into numbers it gives tokes to each & every text
tokenizer.fit_on_texts(train_data["review"])
x_train = pad_sequences(tokenizer.texts_to_sequences(train_data["review"]),maxlen=200)
x_test = pad_sequences(tokenizer.texts_to_sequences(test_data["review"]),maxlen=200)
```

```python
print(x_train)
```

```python
print(x_test)
```

```python
y_train = train_data["sentiment"]
y_test = test_data["sentiment"]
```

```python
y_train
```

```python
y_test
```

```python
y_train = y_train.astype("int32")
y_test = y_test.astype("int32")
```

### LSTM (LONG SHORT-TERM MEMORY)

```python
# building the model

model = Sequential()
model.add(Embedding(input_dim=5000,output_dim=128,input_length=200))
model.add(LSTM(128,dropout=0.2,recurrent_dropout=0.2))
model.add(Dense(1,activation="sigmoid"))
```

```python
# compile the model
model.compile(optimizer = "adam" , loss="binary_crossentropy",metrics=["accuracy"])
```

Training the model

```python
model.fit(x_train,y_train,epochs=5,batch_size=64,validation_split=0.2)
```

Model Evaluation

```python
loss,accuracy = model.evaluate(x_test,y_test)#now here in x_test it will predict the data and compare with actual data
print(f"Test Loss: {loss}")
print(f"Test Accuracy: {accuracy}")
```

Building the predictive system

```python
def predict_sentiment(review):
    sequence = tokenizer.texts_to_sequences([review])#tokenizes text data 
    padded_sequence = pad_sequences(sequence,maxlen=200)# then padding it
    prediction = model.predict(padded_sequence) # this will give a probability
    sentiment = "positive" if prediction[0][0] > 0.5 else "negative"
    return sentiment 
```

```python
# example usage
new_review = "this movie was fantastic. I loved it."
sentiment = predict_sentiment(new_review)
print(f"The sentiment of the review is:  {sentiment}")
```

```python
new_review = "this movie was not that good."
sentiment = predict_sentiment(new_review)
print(f"The sentiment of the review is:  {sentiment}")
```
